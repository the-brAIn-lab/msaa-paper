## 1. Load the locked K=46 fit

```python

def to_float_array(x):
    return np.asarray(x, dtype=float)


def load_msaa_npz(path):
    data = np.load(path, allow_pickle=True)
    results_subj = data["results_subj"].tolist()
    if isinstance(results_subj, np.ndarray):
        results_subj = results_subj.tolist()

    out = {
        "K": int(np.asarray(data["K"]).item()),
        "results_subj": results_subj,
        "condition_labels_str": np.asarray(data["condition_labels_str"]).astype(str),
    }
    for key in ["condition_codes", "condition_names", "centers", "widths", "C", "analysis_type", "fit_scope"]:
        if key in data:
            out[key] = data[key]
    return out


if not FIT_FILE.exists():
    raise FileNotFoundError(f"Missing fit file: {FIT_FILE}")

fit = load_msaa_npz(FIT_FILE)
assert fit["K"] == K, (fit["K"], K)
results_subj = fit["results_subj"]
condition_labels = np.asarray(fit["condition_labels_str"]).astype(str)

print("K:", fit["K"])
print("N participant entries:", len(results_subj))
display(pd.Series(condition_labels).value_counts().reindex(CONDITIONS).to_frame("n"))

example = results_subj[0]
print("sXC shape:", np.asarray(example["sXC"]).shape)
print("S shape:", np.asarray(example["S"]).shape)
```

## 3. Map HTFA nodes to Yeo-7 networks

```python

lookup_table = {
    "Vis": "Visual",
    "SomMot": "Somatomotor",
    "DorsAttn": "Dorsal attention",
    "SalVentAttn": "Ventral attention",
    "Limbic": "Limbic",
    "Cont": "Frontoparietal",
    "Default": "Default mode",
}

NETWORK_ORDER = [
    "Visual", "Somatomotor", "Dorsal attention", "Ventral attention",
    "Limbic", "Frontoparietal", "Default mode",
]

NETWORK_COLORS = {
    "Visual": "#D7DF23",
    "Somatomotor": "#39B54A",
    "Dorsal attention": "#00A79D",
    "Ventral attention": "#27AAE1",
    "Limbic": "#1C75BC",
    "Frontoparietal": "#92278F",
    "Default mode": "#EE2A7B",
}
network_codes = {name: i + 1 for i, name in enumerate(NETWORK_ORDER)}


def nii2cmu(nifti_file, mask_file=None):
    def fullfact(dims):
        vals = np.asmatrix(range(1, dims[0] + 1)).T
        if len(dims) == 1:
            return vals
        aftervals = np.asmatrix(fullfact(dims[1:]))
        inds = np.asmatrix(np.zeros((np.prod(dims), len(dims))))
        row = 0
        for i in range(aftervals.shape[0]):
            inds[row:(row + len(vals)), 0] = vals
            inds[row:(row + len(vals)), 1:] = np.tile(aftervals[i, :], (len(vals), 1))
            row += len(vals)
        return inds

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        img = nib.load(nifti_file)
        mask = NiftiMasker(mask_strategy="background")
        mask.fit(nifti_file if mask_file is None else mask_file)

    S = img.get_sform()
    Y = np.float32(mask.transform(nifti_file)).copy()
    vmask = np.nonzero(np.array(np.reshape(mask.mask_img_.dataobj, (1, np.prod(mask.mask_img_.shape)), order="C")))[1]
    vox_coords = fullfact(img.shape[0:3])[vmask, ::-1] - 1
    R = np.array(np.dot(vox_coords, S[0:3, 0:3])) + S[:3, 3]
    return {"Y": Y, "R": R}


def rbf(R, center, width):
    return np.exp(-np.sum((R - center) ** 2, axis=1) / width)


def node_labels(centers, widths, networks_cmu):
    labels = []
    for c, w in zip(centers, widths):
        rr = rbf(networks_cmu["R"], c, w)
        label_weights = [
            np.sum(rr[networks_cmu["Y"].ravel() == i])
            for i in range(1, len(NETWORK_ORDER) + 1)
        ]
        labels.append(int(np.argmax(label_weights) + 1))
    return pd.DataFrame({
        "code": labels,
        "Network": [NETWORK_ORDER[i - 1] for i in labels],
    })


if not NILEARN_AVAILABLE:
    raise ImportError("This figure requires nibabel/nilearn to reproduce the HTFA-to-Yeo mapping.")
if not SCHAEFER_TXT.exists():
    raise FileNotFoundError(SCHAEFER_TXT)
if not SCHAEFER_NII.exists():
    raise FileNotFoundError(SCHAEFER_NII)

key = pd.read_csv(
    SCHAEFER_TXT,
    sep="\t",
    header=None,
    names=["id", "name", "x", "y", "z", "t"],
).drop("t", axis=1)
key["network"] = key["name"].apply(lambda x: x.split("_")[2])
key["network"] = key["network"].apply(lambda x: lookup_table[x])
key["code"] = key["network"].apply(lambda x: network_codes[x])
key.set_index("id", inplace=True)
key.loc[0, "code"] = 0

networks_cmu = nii2cmu(str(SCHAEFER_NII))
networks_cmu["Y"] = np.atleast_2d(
    np.array([key.loc[i, "code"] for i in networks_cmu["Y"]]).astype(float)
)
node_codes = node_labels(centers, widths, networks_cmu)

print("Yeo node counts:")
display(node_codes["Network"].value_counts().reindex(NETWORK_ORDER).to_frame("n_nodes"))

# ============================================================
# Atlas-mapping audit for the 700 HTFA nodes
#
# Paste AFTER the notebook cell that defines:
#   centers, widths, SCHAEFER_NII, key, NETWORK_ORDER,
#   network_codes, networks_cmu, node_codes, rbf, OUTPUT_DIR
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nibabel as nib

from scipy.spatial import cKDTree
from nibabel.affines import apply_affine


AUDIT_CONFIDENCE_THRESHOLD = 0.50
AUDIT_PREFIX = "Figure5_HTFA_Yeo_mapping_audit"

OUTPUT_DIR = Path(OUTPUT_DIR)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

centers_arr = np.asarray(
    centers,
    dtype=float,
)

widths_arr = np.asarray(
    widths,
    dtype=float,
).ravel()

if centers_arr.shape != (700, 3):
    print(
        "WARNING: expected 700 x 3 centers, found",
        centers_arr.shape,
    )

if len(widths_arr) != len(centers_arr):
    raise ValueError(
        "centers/widths mismatch: "
        f"{len(centers_arr)} vs {len(widths_arr)}"
    )


# ------------------------------------------------------------
# 1. Sample the discrete Schaefer atlas at each node center
# ------------------------------------------------------------

atlas_img = nib.load(
    str(SCHAEFER_NII)
)

atlas_data = np.asarray(
    atlas_img.dataobj
)

atlas_shape = np.asarray(
    atlas_data.shape[:3],
    dtype=int,
)

# Convert MNI/world coordinates to atlas voxel coordinates.
center_vox_float = apply_affine(
    np.linalg.inv(atlas_img.affine),
    centers_arr,
)

# Nearest-neighbor sampling is appropriate for a discrete atlas.
center_vox_nearest = np.rint(
    center_vox_float
).astype(int)

center_in_bounds = np.all(
    (center_vox_nearest >= 0)
    & (
        center_vox_nearest
        < atlas_shape[None, :]
    ),
    axis=1,
)

center_parcel_id = np.zeros(
    len(centers_arr),
    dtype=int,
)

valid = np.where(
    center_in_bounds
)[0]

if len(valid):
    ijk = center_vox_nearest[valid]

    center_parcel_id[valid] = np.asarray(
        atlas_data[
            ijk[:, 0],
            ijk[:, 1],
            ijk[:, 2],
        ]
    ).astype(int)

# Atlas background is coded as zero.
center_in_atlas = (
    center_in_bounds
    & (center_parcel_id > 0)
)

code_to_network = {
    int(code): network
    for network, code in network_codes.items()
}

center_network_code = np.zeros(
    len(centers_arr),
    dtype=int,
)

center_network_name = np.full(
    len(centers_arr),
    "Outside/background",
    dtype=object,
)

for node_i, parcel_id in enumerate(
    center_parcel_id
):
    if (
        parcel_id <= 0
        or parcel_id not in key.index
    ):
        continue

    code = int(
        key.loc[
            parcel_id,
            "code",
        ]
    )

    center_network_code[node_i] = code

    center_network_name[node_i] = (
        code_to_network.get(
            code,
            "Unknown",
        )
    )


# ------------------------------------------------------------
# 2. Recompute integrated-RBF overlap assignments
#
# This duplicates the logic already used by the notebook,
# but retains the full seven-network overlap profile.
# ------------------------------------------------------------

atlas_xyz = np.asarray(
    networks_cmu["R"],
    dtype=float,
)

atlas_network_code = np.asarray(
    networks_cmu["Y"]
).ravel().astype(int)

if (
    atlas_xyz.ndim != 2
    or atlas_xyz.shape[1] != 3
):
    raise ValueError(
        "Unexpected atlas coordinate shape: "
        f"{atlas_xyz.shape}"
    )

# Distance from each HTFA node center to the closest
# non-background Schaefer atlas voxel.
tree = cKDTree(
    atlas_xyz
)

nearest_atlas_distance_mm, _ = tree.query(
    centers_arr,
    k=1,
)

audit_rows = []

for node_i, (center, width) in enumerate(
    zip(
        centers_arr,
        widths_arr,
    )
):
    safe_width = max(
        float(width),
        np.finfo(float).eps,
    )

    # Use the same RBF definition as the existing notebook.
    rr = rbf(
        atlas_xyz,
        center,
        safe_width,
    )

    overlap = np.asarray(
        [
            rr[
                atlas_network_code == code
            ].sum()
            for code in range(
                1,
                len(NETWORK_ORDER) + 1,
            )
        ],
        dtype=float,
    )

    total_overlap = float(
        overlap.sum()
    )

    if total_overlap > 0:
        overlap_fraction = (
            overlap / total_overlap
        )

        order = np.argsort(
            overlap_fraction
        )[::-1]

        assigned_code = int(
            order[0] + 1
        )

        confidence = float(
            overlap_fraction[order[0]]
        )

        second_fraction = float(
            overlap_fraction[order[1]]
        )

        assignment_margin = (
            confidence - second_fraction
        )

    else:
        overlap_fraction = np.full(
            len(NETWORK_ORDER),
            np.nan,
        )

        assigned_code = 0
        confidence = np.nan
        second_fraction = np.nan
        assignment_margin = np.nan

    existing_code = int(
        node_codes.loc[
            node_i,
            "code",
        ]
    )

    existing_network = str(
        node_codes.loc[
            node_i,
            "Network",
        ]
    )

    # This should always pass. It verifies that the audit
    # exactly reproduces the mapping used in Figure 5.
    if assigned_code != existing_code:
        raise RuntimeError(
            "Recomputed integrated-overlap assignment "
            "does not match node_codes for node "
            f"{node_i}: {assigned_code} vs "
            f"{existing_code}"
        )

    row = {
        "node": node_i,
        "x": float(center[0]),
        "y": float(center[1]),
        "z": float(center[2]),
        "width": float(width),

        "assigned_code_integrated_overlap": (
            assigned_code
        ),
        "assigned_network_integrated_overlap": (
            existing_network
        ),

        # Proportion of the node's total cortical RBF
        # overlap assigned to its best-matching network.
        "assignment_confidence": confidence,

        "second_best_overlap_fraction": (
            second_fraction
        ),

        "assignment_margin": (
            assignment_margin
        ),

        "total_cortical_rbf_overlap": (
            total_overlap
        ),

        "center_voxel_i": int(
            center_vox_nearest[
                node_i,
                0,
            ]
        ),
        "center_voxel_j": int(
            center_vox_nearest[
                node_i,
                1,
            ]
        ),
        "center_voxel_k": int(
            center_vox_nearest[
                node_i,
                2,
            ]
        ),

        "center_in_atlas_bounds": bool(
            center_in_bounds[node_i]
        ),

        "center_parcel_id": int(
            center_parcel_id[node_i]
        ),

        "center_in_cortical_atlas": bool(
            center_in_atlas[node_i]
        ),

        "center_network_code": int(
            center_network_code[node_i]
        ),

        "center_network": str(
            center_network_name[node_i]
        ),

        "center_agrees_with_integrated_assignment": (
            bool(
                center_in_atlas[node_i]
                and center_network_code[node_i]
                == assigned_code
            )
        ),

        "distance_to_nearest_atlas_voxel_mm": float(
            nearest_atlas_distance_mm[
                node_i
            ]
        ),
    }

    # Save the complete seven-network overlap profile.
    for network_i, network in enumerate(
        NETWORK_ORDER
    ):
        slug = (
            network.lower()
            .replace(" ", "_")
            .replace("-", "_")
        )

        row[
            f"overlap_fraction_{slug}"
        ] = float(
            overlap_fraction[network_i]
        )

    audit_rows.append(row)


node_audit_df = pd.DataFrame(
    audit_rows
)

audit_csv = (
    OUTPUT_DIR
    / f"{AUDIT_PREFIX}_nodes.csv"
)

node_audit_df.to_csv(
    audit_csv,
    index=False,
)


# ------------------------------------------------------------
# 3. Compact audit summaries
# ------------------------------------------------------------

inside = node_audit_df[
    node_audit_df[
        "center_in_cortical_atlas"
    ]
]

outside = node_audit_df[
    ~node_audit_df[
        "center_in_cortical_atlas"
    ]
]

agreement_rate = (
    inside[
        "center_agrees_with_integrated_assignment"
    ].mean()
    if len(inside)
    else np.nan
)

summary_df = pd.DataFrame(
    [
        {
            "metric": "n_nodes_total",
            "value": len(
                node_audit_df
            ),
        },
        {
            "metric": (
                "n_centers_inside_cortical_atlas"
            ),
            "value": len(inside),
        },
        {
            "metric": (
                "n_centers_outside_or_background"
            ),
            "value": len(outside),
        },
        {
            "metric": (
                "pct_centers_inside_cortical_atlas"
            ),
            "value": (
                100
                * len(inside)
                / len(node_audit_df)
            ),
        },
        {
            "metric": (
                "center_vs_integrated_assignment_"
                "agreement_pct_inside"
            ),
            "value": (
                100 * agreement_rate
            ),
        },
        {
            "metric": (
                "median_assignment_confidence_all"
            ),
            "value": node_audit_df[
                "assignment_confidence"
            ].median(),
        },
        {
            "metric": (
                "median_assignment_confidence_inside"
            ),
            "value": inside[
                "assignment_confidence"
            ].median(),
        },
        {
            "metric": (
                "median_assignment_confidence_outside"
            ),
            "value": outside[
                "assignment_confidence"
            ].median(),
        },
        {
            "metric": (
                "n_assignment_confidence_below_0.50"
            ),
            "value": int(
                (
                    node_audit_df[
                        "assignment_confidence"
                    ]
                    < 0.50
                ).sum()
            ),
        },
        {
            "metric": (
                "n_assignment_confidence_below_0.60"
            ),
            "value": int(
                (
                    node_audit_df[
                        "assignment_confidence"
                    ]
                    < 0.60
                ).sum()
            ),
        },
        {
            "metric": (
                "median_distance_to_atlas_mm_"
                "outside_centers"
            ),
            "value": outside[
                "distance_to_nearest_atlas_voxel_mm"
            ].median(),
        },
    ]
)

summary_csv = (
    OUTPUT_DIR
    / f"{AUDIT_PREFIX}_summary.csv"
)

summary_df.to_csv(
    summary_csv,
    index=False,
)

print(
    "\nAtlas-mapping audit summary"
)

display(
    summary_df
)

print(
    "\nIntegrated-overlap network counts"
)

display(
    node_audit_df[
        "assigned_network_integrated_overlap"
    ]
    .value_counts()
    .reindex(NETWORK_ORDER)
    .rename("n_nodes")
    .to_frame()
)

print(
    "\nCenter status by integrated-overlap assignment"
)

display(
    pd.crosstab(
        node_audit_df[
            "assigned_network_integrated_overlap"
        ],
        node_audit_df[
            "center_in_cortical_atlas"
        ],
    )
    .reindex(NETWORK_ORDER)
    .rename(
        columns={
            False: (
                "center_outside_or_background"
            ),
            True: "center_inside",
        }
    )
)

print("Saved:", audit_csv)
print("Saved:", summary_csv)
```


## 4. Compute temporal profiles and all-node network composition

For each archetype, the temporal motif is the mean `sXC[:, k]` across participant entries. The spatial map is the mean `S[k, :]` across participant entries. Network composition is computed from the absolute spatial coefficient mass across **all 700 nodes**:

\[
\mathrm{mass}_{k,n}=\frac{\sum_{v\in n}|\overline{S}_{k,v}|}{\sum_v |\overline{S}_{k,v}|}.
\]

The network composition therefore does **not** depend on the display threshold.


```python

network_rows = []
spatial_maps = {}
temporal_summary = {}

for k in SELECTED_ARCHETYPES:
    tc = np.stack([to_float_array(sub["sXC"])[:, k] for sub in results_subj], axis=0)
    mean_tc = tc.mean(axis=0)
    sem_tc = tc.std(axis=0, ddof=1) / np.sqrt(tc.shape[0])
    temporal_summary[k] = {"mean": mean_tc, "sem": sem_tc}

    coeffs = np.stack([to_float_array(sub["S"])[k, :] for sub in results_subj], axis=0)
    group_map = np.nan_to_num(coeffs.mean(axis=0), nan=0.0, posinf=0.0, neginf=0.0)
    spatial_maps[k] = group_map

    abs_map = np.abs(group_map)
    total = abs_map.sum()
    for network in NETWORK_ORDER:
        mask = node_codes["Network"].to_numpy() == network
        mass = abs_map[mask].sum() / total if total > 0 else np.nan
        network_rows.append({
            "archetype": k,
            "network": network,
            "mass_fraction": mass,
            "n_nodes": int(mask.sum()),
        })

network_mass_df = pd.DataFrame(network_rows)
display(network_mass_df[network_mass_df["archetype"] == SELECTED_ARCHETYPES[0]])

# ============================================================
# Outside-atlas / assignment-confidence sensitivity for the
# held-out-selected K=46, top-5 network compositions
#
# Paste AFTER:
#   1. the atlas-mapping audit cell above
#   2. the cell that defines spatial_maps
# ============================================================

SENSITIVITY_CONFIDENCE_THRESHOLD = (
    AUDIT_CONFIDENCE_THRESHOLD
)

SENSITIVITY_PREFIX = (
    "Figure5_Yeo7_composition_sensitivity"
)


# Reconstruct spatial_maps if this cell is run separately.
if "spatial_maps" not in globals():
    spatial_maps = {}

    for k in SELECTED_ARCHETYPES:
        coeffs = np.stack(
            [
                to_float_array(
                    sub["S"]
                )[k, :]
                for sub in results_subj
            ],
            axis=0,
        )

        spatial_maps[k] = np.nan_to_num(
            coeffs.mean(axis=0),
            nan=0.0,
            posinf=0.0,
            neginf=0.0,
        )


assigned_network = node_audit_df[
    "assigned_network_integrated_overlap"
].to_numpy()

center_inside_mask = node_audit_df[
    "center_in_cortical_atlas"
].to_numpy(dtype=bool)

confidence_mask = (
    node_audit_df[
        "assignment_confidence"
    ].to_numpy(dtype=float)
    >= SENSITIVITY_CONFIDENCE_THRESHOLD
)

mask_definitions = {
    # Existing Figure 5 analysis.
    "all_nodes": np.ones(
        len(node_audit_df),
        dtype=bool,
    ),

    # Exclude nodes whose MNI centers fall on
    # Schaefer atlas background.
    "center_inside_atlas": (
        center_inside_mask
    ),

    # Retain nodes whose integrated overlap is at
    # least 50% attributable to one Yeo network.
    "confidence_ge_0.50": (
        confidence_mask
    ),

    # Most conservative audit.
    "center_inside_and_confidence_ge_0.50": (
        center_inside_mask
        & confidence_mask
    ),
}


sensitivity_rows = []

for k in SELECTED_ARCHETYPES:
    group_map = np.asarray(
        spatial_maps[k],
        dtype=float,
    )

    if len(group_map) != len(node_audit_df):
        raise ValueError(
            f"Archetype {LABEL[k]}: map has "
            f"{len(group_map)} nodes; audit has "
            f"{len(node_audit_df)}."
        )

    abs_map = np.abs(
        group_map
    )

    full_total = float(
        abs_map.sum()
    )

    for mode, include_mask in (
        mask_definitions.items()
    ):
        include_mask = np.asarray(
            include_mask,
            dtype=bool,
        )

        retained_total = float(
            abs_map[
                include_mask
            ].sum()
        )

        retained_weight_fraction = (
            retained_total / full_total
            if full_total > 0
            else np.nan
        )

        for network in NETWORK_ORDER:
            network_mask = (
                assigned_network
                == network
            )

            use = (
                include_mask
                & network_mask
            )

            mass_fraction = (
                float(
                    abs_map[use].sum()
                    / retained_total
                )
                if retained_total > 0
                else np.nan
            )

            sensitivity_rows.append(
                {
                    "archetype": int(k),
                    "mode": mode,
                    "network": network,
                    "mass_fraction": (
                        mass_fraction
                    ),
                    "n_nodes_included_total": int(
                        include_mask.sum()
                    ),
                    "n_nodes_included_in_network": int(
                        use.sum()
                    ),
                    "retained_absolute_weight_fraction": (
                        retained_weight_fraction
                    ),
                    "confidence_threshold": (
                        SENSITIVITY_CONFIDENCE_THRESHOLD
                    ),
                }
            )


composition_sensitivity_df = pd.DataFrame(
    sensitivity_rows
)

composition_csv = (
    OUTPUT_DIR
    / f"{SENSITIVITY_PREFIX}_profiles.csv"
)

composition_sensitivity_df.to_csv(
    composition_csv,
    index=False,
)


# ------------------------------------------------------------
# Compare restricted profiles with the all-node profile
# ------------------------------------------------------------

comparison_rows = []

for k in SELECTED_ARCHETYPES:
    cur = composition_sensitivity_df[
        composition_sensitivity_df[
            "archetype"
        ].astype(int)
        == int(k)
    ]

    baseline = (
        cur[
            cur["mode"]
            == "all_nodes"
        ]
        .set_index("network")
        .reindex(NETWORK_ORDER)
    )

    base_values = baseline[
        "mass_fraction"
    ].to_numpy(dtype=float)

    for mode in mask_definitions:
        profile = (
            cur[
                cur["mode"] == mode
            ]
            .set_index("network")
            .reindex(NETWORK_ORDER)
        )

        values = profile[
            "mass_fraction"
        ].to_numpy(dtype=float)

        delta = (
            values - base_values
        )

        valid = (
            np.isfinite(values)
            & np.isfinite(base_values)
        )

        if valid.sum() >= 2:
            correlation = float(
                np.corrcoef(
                    base_values[valid],
                    values[valid],
                )[0, 1]
            )
        else:
            correlation = np.nan

        comparison_rows.append(
            {
                "archetype": int(k),
                "mode": mode,

                "n_nodes_included": int(
                    profile[
                        "n_nodes_included_total"
                    ].iloc[0]
                ),

                "retained_absolute_weight_fraction": float(
                    profile[
                        "retained_absolute_weight_fraction"
                    ].iloc[0]
                ),

                "profile_correlation_with_all_nodes": (
                    correlation
                ),

                "l1_profile_difference": float(
                    np.nansum(
                        np.abs(delta)
                    )
                ),

                "max_absolute_network_change": float(
                    np.nanmax(
                        np.abs(delta)
                    )
                ),

                "network_with_largest_change": (
                    NETWORK_ORDER[
                        int(
                            np.nanargmax(
                                np.abs(delta)
                            )
                        )
                    ]
                    if np.isfinite(delta).any()
                    else None
                ),
            }
        )


composition_comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_csv = (
    OUTPUT_DIR
    / f"{SENSITIVITY_PREFIX}_summary.csv"
)

composition_comparison_df.to_csv(
    comparison_csv,
    index=False,
)

print(
    "\nComposition-sensitivity summary"
)

display(
    composition_comparison_df[
        composition_comparison_df[
            "mode"
        ]
        != "all_nodes"
    ]
)


# ------------------------------------------------------------
# Plot all-node versus restricted composition profiles
# ------------------------------------------------------------

modes_to_plot = [
    "all_nodes",
    "center_inside_atlas",
    "confidence_ge_0.50",
]

fig, axes = plt.subplots(
    len(SELECTED_ARCHETYPES),
    1,
    figsize=(
        11,
        2.4 * len(SELECTED_ARCHETYPES),
    ),
    sharex=True,
    constrained_layout=False,
)

if len(SELECTED_ARCHETYPES) == 1:
    axes = [axes]

x = np.arange(
    len(NETWORK_ORDER)
)

for ax, k in zip(
    axes,
    SELECTED_ARCHETYPES,
):
    cur = composition_sensitivity_df[
        composition_sensitivity_df[
            "archetype"
        ].astype(int)
        == int(k)
    ]

    for mode in modes_to_plot:
        profile = (
            cur[
                cur["mode"] == mode
            ]
            .set_index("network")
            .reindex(NETWORK_ORDER)
        )

        ax.plot(
            x,
            profile[
                "mass_fraction"
            ].to_numpy(),
            marker="o",
            linewidth=1.5,
            label=mode.replace(
                "_",
                " ",
            ),
        )

    ax.set_ylabel(
        f"{LABEL[k]}\nMass"
    )

    ax.set_ylim(
        bottom=0
    )

    ax.grid(
        axis="y",
        alpha=0.20,
    )


axes[0].legend(
    frameon=False,
    ncol=3,
    loc="upper center",
    bbox_to_anchor=(
        0.5,
        1.38,
    ),
)

axes[-1].set_xticks(
    x
)

axes[-1].set_xticklabels(
    NETWORK_ORDER,
    rotation=35,
    ha="right",
)

fig.suptitle(
    "Yeo-7 composition sensitivity to atlas-center "
    "and assignment-confidence restrictions",
    fontsize=14,
    fontweight="bold",
    y=0.995,
)

fig.tight_layout(
    rect=[
        0,
        0,
        1,
        0.97,
    ]
)

sensitivity_png = (
    OUTPUT_DIR
    / f"{SENSITIVITY_PREFIX}.png"
)

sensitivity_pdf = (
    OUTPUT_DIR
    / f"{SENSITIVITY_PREFIX}.pdf"
)

fig.savefig(
    sensitivity_png,
    dpi=300,
    bbox_inches="tight",
)

fig.savefig(
    sensitivity_pdf,
    bbox_inches="tight",
)

plt.show()


print("Saved:", composition_csv)
print("Saved:", comparison_csv)
print("Saved:", sensitivity_png)
print("Saved:", sensitivity_pdf)


# ------------------------------------------------------------
# Compact decision-oriented report
# ------------------------------------------------------------

restricted = composition_comparison_df[
    composition_comparison_df[
        "mode"
    ]
    != "all_nodes"
]

print(
    "\nLargest sensitivity observed across "
    "selected archetypes"
)

print(
    "Maximum absolute change in any one "
    "network mass fraction:",
    restricted[
        "max_absolute_network_change"
    ].max(),
)

print(
    "Minimum profile correlation with the "
    "all-node result:",
    restricted[
        "profile_correlation_with_all_nodes"
    ].min(),
)

print(
    "Minimum retained absolute archetype weight:",
    restricted[
        "retained_absolute_weight_fraction"
    ].min(),
)
```

## 5. Cross-fitted top-5 selection frequency

```python

def parse_selected_archetypes(x):
    if isinstance(x, (list, tuple, np.ndarray)):
        return [int(v) for v in x]
    return [int(v) for v in pyast.literal_eval(str(x))]


def selection_frequency_from_runs(run_csv, K, top_m):
    runs = pd.read_csv(run_csv)
    sub = runs[
        (runs["K"].astype(int) == int(K))
        & (runs["top_m"].astype(int) == int(top_m))
        & (runs["condition"].astype(str).isin(CONDITIONS))
    ].copy()

    if len(sub) == 0:
        raise ValueError(f"No K={K} top-{top_m} rows found in {run_csv}")

    rows = []
    for condition in CONDITIONS:
        cur = sub[sub["condition"].astype(str) == condition]
        n_crossfits = len(cur)
        counts = {k: 0 for k in range(K)}
        for selected in cur["selected_archetypes"]:
            for k in parse_selected_archetypes(selected):
                counts[int(k)] += 1
        for k in range(K):
            rows.append({
                "condition": condition,
                "archetype": k,
                "selected_count": counts[k],
                "n_crossfits": n_crossfits,
                "selection_frequency": counts[k] / n_crossfits if n_crossfits > 0 else np.nan,
            })
    return pd.DataFrame(rows)


if HELDOUT_RUN_CSV.exists():
    selection_freq_df = selection_frequency_from_runs(HELDOUT_RUN_CSV, K=K, top_m=TOP_M)
    print("Selection frequency derived from:", HELDOUT_RUN_CSV)
elif HELDOUT_STABILITY_CSV.exists():
    stab = pd.read_csv(HELDOUT_STABILITY_CSV)
    selection_freq_df = stab[(stab["K"].astype(int) == K) & (stab["top_m"].astype(int) == TOP_M)].copy()
    print("Selection frequency loaded from:", HELDOUT_STABILITY_CSV)
else:
    raise FileNotFoundError("Could not find held-out run or stability CSV.")

selected_stability = selection_freq_df[
    selection_freq_df["archetype"].astype(int).isin(SELECTED_ARCHETYPES)
].copy()

pivot_stability = (
    selected_stability
    .pivot_table(index="archetype", columns="condition", values="selection_frequency", aggfunc="mean")
    .reindex(SELECTED_ARCHETYPES)
)
pivot_stability["shared_intact_word"] = pivot_stability[["intact", "word"]].min(axis=1)
display(pivot_stability)
```

This layout keeps the spatial maps readable while making cross-archetype comparisons easier.


```python

from PIL import Image, ImageChops
import io

def scale_node_sizes(values, size_min=18, size_max=105):
    values = np.asarray(values, dtype=float)
    if len(values) == 0:
        return np.array([])
    vmin, vmax = values.min(), values.max()
    if np.isclose(vmax, vmin):
        return np.full(len(values), (size_min + size_max) / 2)
    scaled = (values - vmin) / (vmax - vmin)
    return size_min + scaled * (size_max - size_min)

def render_glass_brain_view(
    coords, networks, sizes, view, cmap_lookup=None
):
    if cmap_lookup is None:
        cmap_lookup = NETWORK_COLORS

    tmp_fig = plt.figure(figsize=(4.0, 3.4))
    display = niplot.plot_glass_brain(
        None,
        display_mode=view,
        figure=tmp_fig,
        black_bg=False,
        colorbar=False,
        plot_abs=False,
        annotate=True,
    )

    for network in NETWORK_ORDER:
        mask = networks == network
        if np.any(mask):
            display.add_markers(
                marker_coords=coords[mask],
                marker_color=cmap_lookup[network],
                marker_size=sizes[mask].tolist(),
            )

    buf = io.BytesIO()
    tmp_fig.savefig(
        buf,
        format="png",
        dpi=220,
        facecolor="white",
        bbox_inches="tight",
        pad_inches=0.02,
    )
    plt.close(tmp_fig)
    buf.seek(0)
    img = Image.open(buf).convert("RGB")

    white = Image.new("RGB", img.size, "white")
    diff = ImageChops.difference(img, white).convert("L")
    diff = diff.point(lambda p: 255 if p > 8 else 0)
    bbox = diff.getbbox()
    if bbox is not None:
        img = img.crop(bbox)

    arr = np.asarray(img)
    buf.close()
    return arr

composition_table = (
    network_mass_df
    .pivot_table(
        index="archetype",
        columns="network",
        values="mass_fraction",
        aggfunc="first",
    )
    .reindex(index=SELECTED_ARCHETYPES, columns=NETWORK_ORDER)
)

stability_heatmap = (
    selected_stability
    .pivot_table(
        index="archetype",
        columns="condition",
        values="selection_frequency",
        aggfunc="mean",
    )
    .reindex(index=SELECTED_ARCHETYPES, columns=CONDITIONS)
)

fig = plt.figure(figsize=(17, 22))

outer = fig.add_gridspec(
    2, 1,
    height_ratios=[6.4, 1.8],
    hspace=0.22,
)

top = outer[0].subgridspec(
    len(SELECTED_ARCHETYPES),
    4,
    width_ratios=[1.55, 1.15, 0.90, 1.15],
    wspace=0.06,
    hspace=0.34,
)

for row_i, k in enumerate(SELECTED_ARCHETYPES):
    ax_t = fig.add_subplot(top[row_i, 0])

    mean_tc = temporal_summary[k]["mean"]
    sem_tc = temporal_summary[k]["sem"]
    t = np.arange(len(mean_tc))

    ax_t.plot(t, mean_tc, linewidth=1.8)
    ax_t.fill_between(
        t, mean_tc - sem_tc, mean_tc + sem_tc, alpha=0.20
    )
    ax_t.axhline(0, linewidth=0.7, linestyle=":")
    ax_t.set_xlim(0, len(mean_tc) - 1)
    ax_t.set_ylabel(
        LABEL[k], rotation=0, labelpad=28,
        fontsize=14, fontweight="bold", va="center"
    )
    ax_t.tick_params(axis="both", labelsize=12)

    if row_i == len(SELECTED_ARCHETYPES) - 1:
        ax_t.set_xlabel("Timepoint", fontsize=14)
    else:
        ax_t.set_xticklabels([])

    if row_i == 0:
        ax_t.set_title(
            "Temporal motif", fontweight="bold", fontsize=15, pad=10
        )

    group_map = spatial_maps[k]
    keep = group_map >= DISPLAY_THR_FRAC * np.max(group_map)
    coords = centers[keep]
    vals = group_map[keep]
    nets = node_codes.loc[keep, "Network"].to_numpy()
    sizes = scale_node_sizes(vals)

    brain_axes = []
    for col_i, view in enumerate(["l", "z", "r"], start=1):
        ax_b = fig.add_subplot(top[row_i, col_i])
        brain_axes.append(ax_b)

        brain_img = render_glass_brain_view(
            coords=coords,
            networks=nets,
            sizes=sizes,
            view=view,
            cmap_lookup=NETWORK_COLORS,
        )
        ax_b.imshow(brain_img, interpolation="antialiased")
        ax_b.set_axis_off()
        ax_b.set_anchor("C")

    if row_i == 0:
        brain_axes[1].set_title(
            "Spatial deployment",
            fontweight="bold",
            fontsize=15,
            pad=10,
        )

bottom = outer[1].subgridspec(
    1, 2,
    width_ratios=[1.70, 1.00],
    wspace=0.18,
)

# Yeo composition
ax_comp = fig.add_subplot(bottom[0, 0])
ypos = np.arange(len(SELECTED_ARCHETYPES))
left = np.zeros(len(SELECTED_ARCHETYPES))

for network in NETWORK_ORDER:
    vals = composition_table[network].to_numpy(dtype=float)
    ax_comp.barh(
        ypos, vals, left=left,
        color=NETWORK_COLORS[network],
        edgecolor="none",
        height=0.72,
        label=network,
    )
    left += vals

ax_comp.set_yticks(ypos)
ax_comp.set_yticklabels(
    [LABEL[k] for k in SELECTED_ARCHETYPES], fontsize=15
)
ax_comp.invert_yaxis()
ax_comp.set_xlim(0, 1)
ax_comp.set_xlabel(
    "Fraction of |spatial weight|", fontsize=15, labelpad=8
)
ax_comp.tick_params(axis="x", labelsize=15)
ax_comp.set_title(
    "Yeo-7 composition across selected archetypes",
    fontweight="bold",
    fontsize=17,
)

# Selection stability
ax_heat = fig.add_subplot(bottom[0, 1])
heat = stability_heatmap.to_numpy(dtype=float)
im = ax_heat.imshow(heat, aspect="auto", vmin=0, vmax=1)

ax_heat.set_xticks(np.arange(len(CONDITIONS)))
ax_heat.set_xticklabels(["Intact", "Word", "Rest"], fontsize=15)
ax_heat.set_yticks(np.arange(len(SELECTED_ARCHETYPES)))
ax_heat.set_yticklabels(
    [LABEL[k] for k in SELECTED_ARCHETYPES], fontsize=15
)
ax_heat.set_title(
    "Cross-fitted top-5 selection stability",
    fontweight="bold",
    fontsize=17,
)

for r in range(heat.shape[0]):
    for c in range(heat.shape[1]):
        val = heat[r, c]
        txt_color = "white" if val >= 0.55 else "black"
        ax_heat.text(
            c, r, f"{val:.2f}",
            ha="center", va="center",
            fontsize=13, color=txt_color
        )

cbar = fig.colorbar(im, ax=ax_heat, fraction=0.046, pad=0.04)
cbar.set_label("Selection frequency", fontsize=15)
cbar.ax.tick_params(labelsize=14)

legend_handles = [
    plt.Rectangle((0, 0), 1, 1, color=NETWORK_COLORS[n])
    for n in NETWORK_ORDER
]
fig.legend(
    legend_handles,
    NETWORK_ORDER,
    loc="lower center",
    ncol=4,
    frameon=False,
    fontsize=14,
    bbox_to_anchor=(0.5, 0.018),
)

fig.suptitle(
    f"Held-out-selected spatial archetypes at K={K}",
    fontsize=19,
    fontweight="bold",
    y=0.995,
)
fig.subplots_adjust(bottom=0.115, top=0.965)

fig.savefig(FIGURE_PNG, dpi=DPI, bbox_inches="tight")
fig.savefig(FIGURE_PDF, bbox_inches="tight")
plt.show()

print("Saved:", FIGURE_PNG)
print("Saved:", FIGURE_PDF)
```
