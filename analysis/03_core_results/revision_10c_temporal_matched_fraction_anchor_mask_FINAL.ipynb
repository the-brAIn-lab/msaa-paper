This is a lightweight follow-on to `revision_10a_temporal_K14_local_reuse.ipynb`.
It reuses the completed canonical fits, participant-held-out rankings, and
unmasked \(K=14\) accumulation results. It adds the two controls needed to
interpret the low-model-order temporal-MS-AA result:

1. **Matched retained-fraction analysis across temporal model order.** For each
   \(K\), select approximately 25%, 50%, 70%, or 100% of the archetypes using
   the participant-held-out ranking, and evaluate on the held-out participant
   half. Exact existing \((K,m)\) cells and stored full-model evaluations are
   reused whenever available.

2. **Generator-anchor-masked \(K=14\) reconstruction decoding.** For each of the
   14 temporal-MS-AA archetypes, define its primary generator TR as the row of
   the shared generator matrix \(C\) with maximum weight. Exclude a common
   window around all 14 generator TRs from both the query and candidate axes of
   the timepoint-matching matrix. The original participant-held-out rankings
   are frozen, so this is a conservative test of whether the previously
   selected subsets generalize away from their defining timepoints.

The masked decoder uses the same outer participant splits and inner decoder
seeds as the unmasked analysis. The time mask is identical for every \(m\),
and reconstructions are renormalized after masking. The resulting chance level
is \(1/T_{\mathrm{remaining}}\).

## 1. Discover and load temporal MS-AA fits

Copied fits are deduplicated by a fingerprint of the shared generator matrix
\(C\). The canonical repository fit is listed first when available.


````python
# ============================================================
# FIT DISCOVERY / LOADING
# ============================================================

FIT_NAME_TEMPLATE = "temporalAA_across_acrossCond_K{K}.npz"


def normalize_labels(values):
    out = []
    for value in np.asarray(values).reshape(-1):
        out.append(value.decode("utf-8") if isinstance(value, bytes) else str(value))
    return np.asarray(out, dtype=object)


def load_fit(path):
    path = Path(path)
    data = np.load(path, allow_pickle=True)
    results_subj = data["results_subj"].tolist()
    if isinstance(results_subj, np.ndarray):
        results_subj = results_subj.tolist()
    return {
        "path": path,
        "K": int(np.asarray(data["K"]).item()),
        "results_subj": results_subj,
        "condition_labels_str": normalize_labels(data["condition_labels_str"]),
        "C": np.asarray(data["C"], dtype=float) if "C" in data else None,
    }


def fit_fingerprint(path):
    fit = load_fit(path)
    digest = hashlib.sha1()
    if fit["C"] is not None:
        digest.update(np.round(fit["C"], 10).tobytes())
    else:
        digest.update(np.round(np.asarray(fit["results_subj"][0]["sXC"], dtype=float), 10).tobytes())
    return digest.hexdigest()[:16]


def seed_label_from_path(path, fallback_index):
    match = re.search(r"seed[_-]?(\d+)", str(path), flags=re.IGNORECASE)
    if match:
        return f"seed{int(match.group(1))}"
    canonical = CANONICAL_FIT_DIR / FIT_NAME_TEMPLATE.format(K=REFERENCE_K)
    try:
        if Path(path).resolve() == canonical.resolve():
            return "seed0"
    except Exception:
        pass
    return f"fit{fallback_index:02d}"


def discover_fit_paths(K):
    filename = FIT_NAME_TEMPLATE.format(K=int(K))
    candidates = []
    canonical = CANONICAL_FIT_DIR / filename
    if canonical.exists():
        candidates.append(canonical)
    if int(K) == REFERENCE_K:
        candidates.extend(Path(p) for p in EXPLICIT_K14_FITS if Path(p).exists())
    if AUTO_DISCOVER_K14_SEEDS or int(K) != REFERENCE_K:
        for root in FIT_SEARCH_ROOTS:
            root = Path(root)
            if not root.exists():
                continue
            if root.is_file():
                if root.name == filename:
                    candidates.append(root)
                continue
            direct = root / filename
            if direct.exists():
                candidates.append(direct)
            try:
                candidates.extend(root.glob(f"**/{filename}"))
            except OSError:
                pass

    rows, seen_paths, seen_fingerprints = [], set(), set()
    for path in candidates:
        try:
            resolved = path.resolve()
        except Exception:
            resolved = path
        if resolved in seen_paths:
            continue
        seen_paths.add(resolved)
        try:
            fit = load_fit(path)
            if fit["K"] != int(K):
                continue
            fp = fit_fingerprint(path)
        except Exception as exc:
            warnings.warn(f"Skipping unreadable fit {path}: {exc}")
            continue
        if fp in seen_fingerprints:
            continue
        seen_fingerprints.add(fp)
        rows.append({
            "fit_path": str(path),
            "fit_label": seed_label_from_path(path, len(rows)),
            "fingerprint": fp,
            "K": int(K),
            "n_observations": len(fit["results_subj"]),
        })

    table = pd.DataFrame(rows)
    if len(table):
        table["_canonical"] = table["fit_path"].astype(str).eq(str(canonical)).astype(int)
        table = table.sort_values(["_canonical", "fit_label"], ascending=[False, True]).drop(columns="_canonical").reset_index(drop=True)
    return table


def condition_subjects_from_fit(fit, condition):
    idx = np.where(fit["condition_labels_str"] == str(condition))[0]
    return [fit["results_subj"][int(i)] for i in idx]


k14_fit_table = discover_fit_paths(REFERENCE_K)
if len(k14_fit_table) == 0:
    raise FileNotFoundError("No K=14 temporal across-condition fits found. Add paths to EXPLICIT_K14_FITS or FIT_SEARCH_ROOTS.")

display(k14_fit_table)
K14_FIT_INDICES = list(range(len(k14_fit_table))) if K14_FIT_INDICES_TO_RUN is None else [int(i) for i in K14_FIT_INDICES_TO_RUN if 0 <= int(i) < len(k14_fit_table)]
print("K=14 fit indices:", K14_FIT_INDICES)
````

## 2. Optimized temporal held-out decoder

This adapts Ben's low-memory temporal cross-fit implementation. Single-archetype
ranking uses an exact low-rank correlation calculation, while cumulative
held-out reconstructions are built incrementally.


````python
# ============================================================
# DECODING CORE
# ============================================================


def get_xval_assignments(ndata, nfolds, rng):
    assignments = np.zeros(ndata, dtype=int)
    groupsize = int(np.ceil(ndata / nfolds))
    for fold in range(1, nfolds):
        inds = np.arange(fold * groupsize, min((fold + 1) * groupsize, ndata))
        assignments[inds] = fold
    rng.shuffle(assignments)
    return assignments


def decoder_metrics(corrs):
    T = corrs.shape[0]
    true_t = np.arange(T)
    pred = np.argmax(corrs, axis=1)
    return {
        "accuracy": float(np.mean(pred == true_t)),
        "error": float(np.mean(np.abs(pred - true_t)) / T),
        "rank": float(np.mean(np.mean(corrs <= np.diag(corrs)[:, None], axis=1))),
    }


def summarize_metric_rows(rows):
    df = pd.DataFrame(rows)
    return {
        "mean": float(df["accuracy"].mean()),
        "err": float(df["accuracy"].std(ddof=1) / np.sqrt(max(len(df), 1))),
        "std": float(df["accuracy"].std(ddof=1)),
        "count": int(len(df)),
        "rank_mean": float(df["rank"].mean()),
        "error_mean": float(df["error"].mean()),
    }


def inner_masks(n_subjects, nreps, seed):
    if NFOLDS_DECODE != 2:
        raise ValueError("Optimized decoder assumes NFOLDS_DECODE=2.")
    rng = np.random.default_rng(int(seed))
    masks = np.zeros((nreps, n_subjects), dtype=bool)
    for rep in range(nreps):
        masks[rep] = get_xval_assignments(n_subjects, 2, rng) == 0
    return masks


def batch_decode_stack(stack, nreps=20, seed=42):
    X = np.asarray(stack, dtype=float)
    N, T, F = X.shape
    if F < 2:
        return {"mean":np.nan,"err":np.nan,"std":np.nan,"count":0,"rank_mean":np.nan,"error_mean":np.nan}
    masks0 = inner_masks(N, nreps, seed)
    n0 = masks0.sum(axis=1).astype(float)
    n1 = (N - masks0.sum(axis=1)).astype(float)
    A = np.einsum("rn,ntf->rtf", masks0 / n0[:, None], X, optimize=True)
    B = np.einsum("rn,ntf->rtf", (~masks0) / n1[:, None], X, optimize=True)
    A = A - A.mean(axis=2, keepdims=True)
    B = B - B.mean(axis=2, keepdims=True)
    normA = np.linalg.norm(A, axis=2)
    normB = np.linalg.norm(B, axis=2)
    A = A / np.where(normA[:, :, None] > 0, normA[:, :, None], 1.0)
    B = B / np.where(normB[:, :, None] > 0, normB[:, :, None], 1.0)
    corrs = np.matmul(A, np.swapaxes(B, 1, 2))
    corrs = np.nan_to_num(corrs, nan=0.0, posinf=0.0, neginf=0.0)
    rows=[]
    for rep in range(nreps):
        m0=decoder_metrics(corrs[rep]); m0.update({"rep":rep,"fold":0}); rows.append(m0)
        m1=decoder_metrics(corrs[rep].T); m1.update({"rep":rep,"fold":1}); rows.append(m1)
    return summarize_metric_rows(rows)


def temporal_single_factors(subjects, archetype):
    U, W = [], []
    for sub in subjects:
        sXC = np.asarray(sub["sXC"], dtype=float)
        S = np.asarray(sub["S"], dtype=float)
        temporal = S[int(archetype), :]
        spatial = sXC[:, int(archetype)]
        centered_t = temporal - temporal.mean()
        sd_t = temporal.std()
        U.append(centered_t)
        W.append(spatial / (np.abs(spatial) * sd_t + EPS))
    U = np.asarray(U, dtype=float)
    W = np.asarray(W, dtype=float)
    Wc = W - W.mean(axis=1, keepdims=True)
    return U, Wc @ Wc.T


def lowrank_single_decode_summary(U, G, subset_indices, nreps, seed):
    subset_indices = np.asarray(subset_indices, dtype=int)
    rng = np.random.default_rng(int(seed))
    rows=[]
    for rep in range(nreps):
        fold_ids = get_xval_assignments(len(subset_indices), 2, rng)
        ia = subset_indices[fold_ids == 0]
        ib = subset_indices[fold_ids == 1]
        Ua, Ub = U[ia].T, U[ib].T
        Gaa = G[np.ix_(ia, ia)] / (len(ia) ** 2)
        Gbb = G[np.ix_(ib, ib)] / (len(ib) ** 2)
        Gab = G[np.ix_(ia, ib)] / (len(ia) * len(ib))
        numerator = Ua @ Gab @ Ub.T
        na = np.sqrt(np.maximum(np.einsum("ti,ij,tj->t", Ua, Gaa, Ua, optimize=True), 0.0))
        nb = np.sqrt(np.maximum(np.einsum("ti,ij,tj->t", Ub, Gbb, Ub, optimize=True), 0.0))
        denom = na[:, None] * nb[None, :]
        corrs = np.divide(numerator, denom, out=np.zeros_like(numerator), where=denom > 0)
        corrs = np.nan_to_num(corrs, nan=0.0, posinf=0.0, neginf=0.0)
        m0=decoder_metrics(corrs); m0.update({"rep":rep,"fold":0}); rows.append(m0)
        m1=decoder_metrics(corrs.T); m1.update({"rep":rep,"fold":1}); rows.append(m1)
    return summarize_metric_rows(rows)
````

````python
# ============================================================
# RECONSTRUCTION / EXPRESSION BUILDERS
# ============================================================


def temporal_contribution_stack(subjects, archetype):
    tvec = np.stack([np.asarray(sub["S"], dtype=float)[int(archetype), :] for sub in subjects], axis=0)
    vvec = np.stack([np.asarray(sub["sXC"], dtype=float)[:, int(archetype)] for sub in subjects], axis=0)
    return tvec[:, :, None] * vvec[:, None, :]


def normalize_stack_over_time(raw):
    return (raw - raw.mean(axis=1, keepdims=True)) / (raw.std(axis=1, keepdims=True) + EPS)


def full_temporal_reconstruction_stack(subjects):
    out=[]
    for sub in subjects:
        X = (np.asarray(sub["sXC"], dtype=float) @ np.asarray(sub["S"], dtype=float)).T
        out.append((X - X.mean(axis=0, keepdims=True)) / (X.std(axis=0, keepdims=True) + EPS))
    return np.stack(out, axis=0)


def temporal_expression_stack(subjects, indices=None):
    out=[]
    for sub in subjects:
        S=np.asarray(sub["S"], dtype=float)
        idx=np.arange(S.shape[0]) if indices is None else np.asarray(list(indices), dtype=int)
        X=S[idx, :].T
        out.append((X - X.mean(axis=0, keepdims=True)) / (X.std(axis=0, keepdims=True) + EPS))
    return np.stack(out, axis=0)


def evaluate_reconstruction_topm_fast(subjects, ranked, top_m_values, nreps, seed):
    K=len(ranked)
    desired=sorted({int(m) for m in top_m_values if 1 <= int(m) <= K})
    if not desired:
        return pd.DataFrame()
    full=full_temporal_reconstruction_stack(subjects)
    full_summary=batch_decode_stack(full, nreps=nreps, seed=seed)
    T,V=full.shape[1],full.shape[2]
    chance=1.0/T
    raw=np.zeros((len(subjects),T,V), dtype=float)
    rows=[]
    for pos,arch in enumerate(ranked[:max(desired)], start=1):
        raw += temporal_contribution_stack(subjects, arch)
        if pos not in desired:
            continue
        summary=batch_decode_stack(normalize_stack_over_time(raw), nreps=nreps, seed=seed)
        denom=full_summary["mean"]-chance
        retention=(summary["mean"]-chance)/denom if abs(denom)>1e-12 else np.nan
        rows.append({
            "space":"reconstruction","top_m":pos,"selected_archetypes":str([int(x) for x in ranked[:pos]]),
            "mean":summary["mean"],"err":summary["err"],"std":summary["std"],"count":summary["count"],
            "rank_mean":summary["rank_mean"],"error_mean":summary["error_mean"],
            "full_mean_same_eval":full_summary["mean"],"full_err_same_eval":full_summary["err"],
            "retention_above_chance":retention,"chance":chance,
        })
    return pd.DataFrame(rows)


def evaluate_expression_topm_fast(subjects, ranked, top_m_values, nreps, seed):
    K=len(ranked)
    desired=sorted({int(m) for m in top_m_values if 1 <= int(m) <= K})
    if not desired:
        return pd.DataFrame()
    full=temporal_expression_stack(subjects, None)
    full_summary=batch_decode_stack(full, nreps=nreps, seed=seed)
    chance=1.0/full.shape[1]
    rows=[]
    for m in desired:
        selected=ranked[:m]
        summary=(batch_decode_stack(temporal_expression_stack(subjects, selected), nreps=nreps, seed=seed)
                 if m >= 2 else {"mean":np.nan,"err":np.nan,"std":np.nan,"count":0,"rank_mean":np.nan,"error_mean":np.nan})
        denom=full_summary["mean"]-chance
        retention=((summary["mean"]-chance)/denom if np.isfinite(summary["mean"]) and abs(denom)>1e-12 else np.nan)
        rows.append({
            "space":"expression","top_m":m,"selected_archetypes":str([int(x) for x in selected]),
            "mean":summary["mean"],"err":summary["err"],"std":summary["std"],"count":summary["count"],
            "rank_mean":summary["rank_mean"],"error_mean":summary["error_mean"],
            "full_mean_same_eval":full_summary["mean"],"full_err_same_eval":full_summary["err"],
            "retention_above_chance":retention,"chance":chance,
        })
    return pd.DataFrame(rows)
````

## 3. Fixed outer splits and reusable rankings

The default split scheme reuses the same participant splits across model orders
and fit seeds, making the matched-fraction comparison paired with respect to
participant assignment.


````python
# ============================================================
# LEGACY SPLITS / SEEDS / CACHE IMPORT
# ============================================================

def outer_seed_for(condition, outer_rep, K):
    # Exact split definition used by the optimized temporal full-grid notebook.
    sequence = np.random.SeedSequence([
        int(RNG_SEED),
        1,  # temporal-analysis tag
        int(K),
        int(CONDITION_INDEX[condition]),
        int(outer_rep),
    ])
    return int(sequence.generate_state(1, dtype=np.uint32)[0])


def ranking_seed_for(K, outer_rep, direction_offset):
    return int(
        RNG_SEED
        + 100000 * int(outer_rep)
        + 1000 * int(K)
        + int(direction_offset)
    )


def evaluation_seed_for(K, outer_rep, direction_offset):
    return int(
        RNG_SEED
        + 200000 * int(outer_rep)
        + 1000 * int(K)
        + int(direction_offset)
    )


def split_outer_subject_indices(n_subjects, seed):
    rng = np.random.default_rng(int(seed))
    indices = np.arange(n_subjects)
    rng.shuffle(indices)
    midpoint = n_subjects // 2
    return indices[:midpoint], indices[midpoint:]


def direction_definitions(n_subjects, condition, K):
    rows = []
    for outer_rep in range(N_OUTER_REPS):
        outer_seed = outer_seed_for(condition, outer_rep, K)
        group_a, group_b = split_outer_subject_indices(
            n_subjects,
            outer_seed,
        )
        rows.extend([
            {
                "outer_rep": outer_rep,
                "outer_seed": outer_seed,
                "direction": "A_to_B",
                "ranking_idx": group_a,
                "evaluation_idx": group_b,
                "direction_offset": 0,
            },
            {
                "outer_rep": outer_rep,
                "outer_seed": outer_seed,
                "direction": "B_to_A",
                "ranking_idx": group_b,
                "evaluation_idx": group_a,
                "direction_offset": 1,
            },
        ])
    return rows


def load_compatible_legacy_csv(filename):
    frames = []
    sources = []
    for directory in LEGACY_OUTPUT_DIR_CANDIDATES:
        path = Path(directory) / filename
        if not path.exists() or path.stat().st_size == 0:
            continue
        try:
            frame = pd.read_csv(path)
        except Exception as exc:
            warnings.warn(f"Could not read {path}: {exc}")
            continue
        frame["_legacy_source_csv"] = str(path)
        frames.append(frame)
        sources.append(path)

    if not frames:
        return pd.DataFrame(), []

    combined = pd.concat(frames, ignore_index=True, sort=False)

    if "analysis_type" in combined.columns:
        combined = combined[
            combined["analysis_type"].astype(str) == ANALYSIS_TYPE
        ]
    if "fit_scope" in combined.columns:
        combined = combined[
            combined["fit_scope"].astype(str) == FIT_SCOPE
        ]

    return combined.reset_index(drop=True), sources


legacy_rank_raw, legacy_rank_sources = load_compatible_legacy_csv(
    "heldout_per_archetype_ranking_runs.csv"
)
legacy_run_raw, legacy_run_sources = load_compatible_legacy_csv(
    "heldout_topm_crossfit_runs.csv"
)
legacy_seed_raw, legacy_seed_sources = load_compatible_legacy_csv(
    "outer_split_seeds.csv"
)

print("Legacy ranking sources:", [str(p) for p in legacy_rank_sources])
print("Legacy run sources:", [str(p) for p in legacy_run_sources])
print("Legacy ranking rows:", len(legacy_rank_raw))
print("Legacy held-out rows:", len(legacy_run_raw))


RANK_CSV = OUTPUT_DIR / "temporal_crossfit_rankings.csv"
rank_df = load_csv(RANK_CSV)

RANK_KEYS = [
    "fit_fingerprint",
    "condition",
    "K",
    "outer_rep",
    "direction",
    "archetype",
]


def import_legacy_rankings_for_fit(
    fit,
    fit_label,
    fingerprint,
    condition,
):
    global rank_df

    if not REUSE_EXISTING_RANKINGS or len(legacy_rank_raw) == 0:
        return 0

    K = int(fit["K"])

    required = {
        "condition",
        "K",
        "outer_rep",
        "direction",
        "archetype",
    }
    if not required.issubset(legacy_rank_raw.columns):
        warnings.warn(
            "Legacy ranking CSV lacks required columns; rankings will be "
            "computed only where needed."
        )
        return 0

    current = legacy_rank_raw[
        (legacy_rank_raw["condition"].astype(str) == str(condition))
        & (legacy_rank_raw["K"].astype(int) == K)
    ].copy()

    if len(current) == 0:
        return 0

    rename = {}
    if "mean_accuracy" not in current.columns and "mean" in current.columns:
        rename["mean"] = "mean_accuracy"
    if rename:
        current = current.rename(columns=rename)

    for column, default in {
        "mean_accuracy": np.nan,
        "sem_accuracy": np.nan,
        "rank_mean": np.nan,
        "error_mean": np.nan,
        "outer_seed": np.nan,
        "n_ranking_subjects": 18,
    }.items():
        if column not in current.columns:
            current[column] = default

    current["fit_label"] = fit_label
    current["fit_fingerprint"] = fingerprint
    current["fit_path"] = str(fit["path"])

    # Verify or fill the expected split seeds.
    expected_seed = current.apply(
        lambda row: outer_seed_for(
            str(row["condition"]),
            int(row["outer_rep"]),
            int(row["K"]),
        ),
        axis=1,
    )

    if current["outer_seed"].notna().any():
        mismatch = (
            current["outer_seed"].notna()
            & (
                current["outer_seed"].astype("Int64")
                != pd.Series(expected_seed, index=current.index).astype("Int64")
            )
        )
        if mismatch.any():
            raise RuntimeError(
                "Existing ranking rows use a different outer-split seed "
                "scheme. Do not mix them with newly computed rows."
            )

    current["outer_seed"] = expected_seed.astype(int)

    keep = [
        "fit_label",
        "fit_fingerprint",
        "fit_path",
        "condition",
        "K",
        "outer_rep",
        "direction",
        "outer_seed",
        "n_ranking_subjects",
        "archetype",
        "mean_accuracy",
        "sem_accuracy",
        "rank_mean",
        "error_mean",
    ]

    current = current[keep]

    before = len(rank_df)
    rank_df = append_dedupe(
        rank_df,
        current,
        RANK_KEYS,
    )
    save_csv(rank_df, RANK_CSV)
    return len(rank_df) - before


def canonicalize_legacy_runs_for_fit(
    fit,
    fit_label,
    fingerprint,
    condition=None,
):
    if not REUSE_EXISTING_RECONSTRUCTION_RUNS or len(legacy_run_raw) == 0:
        return pd.DataFrame()

    K = int(fit["K"])

    required = {
        "condition",
        "K",
        "outer_rep",
        "direction",
        "top_m",
    }
    if not required.issubset(legacy_run_raw.columns):
        warnings.warn(
            "Legacy run CSV lacks required columns; held-out cells will be "
            "computed only where needed."
        )
        return pd.DataFrame()

    current = legacy_run_raw[
        legacy_run_raw["K"].astype(int) == K
    ].copy()

    if condition is not None:
        current = current[
            current["condition"].astype(str) == str(condition)
        ]

    if len(current) == 0:
        return pd.DataFrame()

    if "mean" not in current.columns and "accuracy" in current.columns:
        current = current.rename(columns={"accuracy": "mean"})

    defaults = {
        "err": np.nan,
        "std": np.nan,
        "count": np.nan,
        "rank_mean": np.nan,
        "error_mean": np.nan,
        "full_mean_same_eval": np.nan,
        "full_err_same_eval": np.nan,
        "chance": 1.0 / 300.0,
        "retention_above_chance": np.nan,
        "selected_archetypes": "[]",
        "outer_seed": np.nan,
        "n_ranking_subjects": 18,
        "n_evaluation_subjects": 18,
    }
    for column, default in defaults.items():
        if column not in current.columns:
            current[column] = default

    missing_retention = current["retention_above_chance"].isna()
    denominator = (
        current["full_mean_same_eval"].astype(float)
        - current["chance"].astype(float)
    )
    current.loc[
        missing_retention & denominator.abs().gt(1e-12),
        "retention_above_chance",
    ] = (
        current.loc[missing_retention, "mean"].astype(float)
        - current.loc[missing_retention, "chance"].astype(float)
    ) / denominator.loc[missing_retention]

    current["space"] = "reconstruction"
    current["fit_label"] = fit_label
    current["fit_fingerprint"] = fingerprint
    current["fit_path"] = str(fit["path"])

    expected_seed = current.apply(
        lambda row: outer_seed_for(
            str(row["condition"]),
            int(row["outer_rep"]),
            int(row["K"]),
        ),
        axis=1,
    )
    if current["outer_seed"].notna().any():
        mismatch = (
            current["outer_seed"].notna()
            & (
                current["outer_seed"].astype("Int64")
                != pd.Series(expected_seed, index=current.index).astype("Int64")
            )
        )
        if mismatch.any():
            raise RuntimeError(
                "Existing held-out rows use a different outer-split seed "
                "scheme. Do not mix them with newly computed rows."
            )
    current["outer_seed"] = expected_seed.astype(int)

    keep = [
        "space",
        "top_m",
        "selected_archetypes",
        "mean",
        "err",
        "std",
        "count",
        "rank_mean",
        "error_mean",
        "full_mean_same_eval",
        "full_err_same_eval",
        "retention_above_chance",
        "chance",
        "fit_label",
        "fit_fingerprint",
        "fit_path",
        "condition",
        "K",
        "outer_rep",
        "direction",
        "outer_seed",
        "n_ranking_subjects",
        "n_evaluation_subjects",
    ]
    return current[keep].reset_index(drop=True)


def ensure_rankings(
    fit,
    fit_label,
    fingerprint,
    condition,
):
    global rank_df

    K = int(fit["K"])
    subjects = condition_subjects_from_fit(fit, condition)
    definitions = direction_definitions(
        len(subjects),
        condition,
        K,
    )

    imported = import_legacy_rankings_for_fit(
        fit,
        fit_label,
        fingerprint,
        condition,
    )
    if imported:
        print(
            f"  imported {imported} existing ranking rows | "
            f"{condition} | K={K}",
            flush=True,
        )

    for archetype in range(K):
        existing = set()
        if len(rank_df):
            current = rank_df[
                (rank_df["fit_fingerprint"].astype(str) == str(fingerprint))
                & (rank_df["condition"].astype(str) == str(condition))
                & (rank_df["K"].astype(int) == K)
                & (rank_df["archetype"].astype(int) == int(archetype))
            ]
            existing = {
                (int(row.outer_rep), str(row.direction))
                for row in current.itertuples(index=False)
            }

        needed = [
            definition
            for definition in definitions
            if (
                int(definition["outer_rep"]),
                str(definition["direction"]),
            ) not in existing
        ]

        if not needed:
            continue

        U, gram = temporal_single_factors(
            subjects,
            archetype,
        )
        rows = []

        for definition in needed:
            summary = lowrank_single_decode_summary(
                U,
                gram,
                definition["ranking_idx"],
                nreps=NREPS_DECODE,
                seed=ranking_seed_for(
                    K,
                    definition["outer_rep"],
                    definition["direction_offset"],
                ),
            )
            rows.append({
                "fit_label": fit_label,
                "fit_fingerprint": fingerprint,
                "fit_path": str(fit["path"]),
                "condition": condition,
                "K": K,
                "outer_rep": int(definition["outer_rep"]),
                "direction": str(definition["direction"]),
                "outer_seed": int(definition["outer_seed"]),
                "n_ranking_subjects": int(len(definition["ranking_idx"])),
                "archetype": int(archetype),
                "mean_accuracy": summary["mean"],
                "sem_accuracy": summary["err"],
                "rank_mean": summary["rank_mean"],
                "error_mean": summary["error_mean"],
            })

        rank_df = append_dedupe(
            rank_df,
            pd.DataFrame(rows),
            RANK_KEYS,
        )

        if (archetype + 1) % 10 == 0 or archetype + 1 == K:
            save_csv(rank_df, RANK_CSV)
            print(
                f"  ranking {fit_label} | {condition} | "
                f"K={K} | {archetype + 1}/{K}",
                flush=True,
            )

        del U
        del gram
        gc.collect()

    save_csv(rank_df, RANK_CSV)
    return definitions, subjects


def ranked_for_direction(
    fingerprint,
    condition,
    K,
    outer_rep,
    direction,
):
    current = rank_df[
        (rank_df["fit_fingerprint"].astype(str) == str(fingerprint))
        & (rank_df["condition"].astype(str) == str(condition))
        & (rank_df["K"].astype(int) == int(K))
        & (rank_df["outer_rep"].astype(int) == int(outer_rep))
        & (rank_df["direction"].astype(str) == str(direction))
    ].copy()

    if current["archetype"].nunique() != int(K):
        raise RuntimeError(
            f"Incomplete ranking: {condition}, K={K}, rep={outer_rep}, "
            f"{direction}: {current['archetype'].nunique()}/{K}"
        )

    return (
        current
        .sort_values(
            ["mean_accuracy", "rank_mean", "error_mean", "archetype"],
            ascending=[False, False, True, True],
        )["archetype"]
        .astype(int)
        .tolist()
    )
````

## 1. Matched retained fractions across temporal model order

This section is inherited from Revision 10A. It first imports compatible exact
\((K,m)\) held-out cells from the established temporal full-grid outputs. It
also treats the stored full-model evaluation as the \(m=K\) result. Only
missing fractions are decoded.


````python
# ============================================================
# MATCHED FRACTIONS — CANONICAL FIT, REUSE EXACT EXISTING CELLS
# ============================================================

FRACTION_RUN_CSV = OUTPUT_DIR / "matched_fraction_runs.csv"
FRACTION_RUN_KEYS = [
    "fit_fingerprint",
    "condition",
    "K",
    "outer_rep",
    "direction",
    "fraction_requested",
    "top_m",
]
fraction_run_df = load_csv(FRACTION_RUN_CSV)


def available_temporal_K():
    values = []
    for root in FIT_SEARCH_ROOTS:
        root = Path(root)
        if not root.exists():
            continue
        try:
            for path in root.glob(
                "**/temporalAA_across_acrossCond_K*.npz"
            ):
                match = re.search(r"_K(\d+)\.npz$", path.name)
                if match:
                    values.append(int(match.group(1)))
        except OSError:
            pass
    return sorted(set(values))


def prefill_fraction_rows_from_existing(
    fit,
    fit_label,
    fingerprint,
    mapping,
):
    existing_runs = canonicalize_legacy_runs_for_fit(
        fit,
        fit_label,
        fingerprint,
    )
    if len(existing_runs) == 0:
        return pd.DataFrame()

    rows = []
    K = int(fit["K"])

    group_columns = [
        "condition",
        "outer_rep",
        "direction",
    ]

    for keys, current in existing_runs.groupby(group_columns):
        condition, outer_rep, direction = keys

        for map_row in mapping.itertuples(index=False):
            requested = float(map_row.fraction_requested)
            top_m = int(map_row.top_m)

            exact = current[
                current["top_m"].astype(int) == top_m
            ]

            if len(exact):
                source = exact.iloc[0].to_dict()
            elif top_m == K and current["full_mean_same_eval"].notna().any():
                # The full model is already evaluated on these exact held-out
                # participants in every existing top-m row.
                base = current[
                    current["full_mean_same_eval"].notna()
                ].iloc[0].to_dict()
                source = base
                source["space"] = "reconstruction"
                source["top_m"] = K
                source["selected_archetypes"] = str(list(range(K)))
                source["mean"] = float(base["full_mean_same_eval"])
                source["err"] = float(base["full_err_same_eval"])
                source["retention_above_chance"] = 1.0
            else:
                continue

            source.update({
                "fraction_requested": requested,
                "actual_fraction": float(map_row.actual_fraction),
                "fit_label": fit_label,
                "fit_fingerprint": fingerprint,
                "fit_path": str(fit["path"]),
                "condition": str(condition),
                "K": K,
                "outer_rep": int(outer_rep),
                "direction": str(direction),
            })
            rows.append(source)

    return pd.DataFrame(rows)


if RUN_MATCHED_FRACTION:
    available = available_temporal_K()
    matched_K = (
        available
        if MATCHED_FRACTION_K_VALUES is None
        else [
            int(K)
            for K in MATCHED_FRACTION_K_VALUES
            if int(K) in available
        ]
    )
    print("Matched-fraction K values:", matched_K)

    for K in matched_K:
        fit_table = discover_fit_paths(K)
        if len(fit_table) == 0:
            continue

        fit_indices = [0]  # canonical/local only

        mapping = pd.DataFrame([
            {
                "fraction_requested": float(fraction),
                "top_m": int(
                    np.clip(
                        np.round(float(fraction) * int(K)),
                        1,
                        int(K),
                    )
                ),
            }
            for fraction in MATCHED_FRACTIONS
        ])
        mapping["actual_fraction"] = (
            mapping["top_m"] / float(K)
        )

        for fit_index in fit_indices:
            fit_row = fit_table.iloc[int(fit_index)]
            fit = load_fit(fit_row["fit_path"])
            fit_label = str(fit_row["fit_label"])
            fingerprint = str(fit_row["fingerprint"])

            prefilled = prefill_fraction_rows_from_existing(
                fit,
                fit_label,
                fingerprint,
                mapping,
            )
            if len(prefilled):
                before = len(fraction_run_df)
                fraction_run_df = append_dedupe(
                    fraction_run_df,
                    prefilled,
                    FRACTION_RUN_KEYS,
                )
                save_csv(fraction_run_df, FRACTION_RUN_CSV)
                print(
                    f"  K={K}: imported "
                    f"{len(fraction_run_df) - before} exact existing cells",
                    flush=True,
                )

            print("\n" + "=" * 88, flush=True)
            print(
                f"Matched fractions | {fit_label} | K={K}",
                flush=True,
            )
            print("=" * 88, flush=True)

            for condition in CONDITIONS:
                definitions, subjects = ensure_rankings(
                    fit,
                    fit_label,
                    fingerprint,
                    condition,
                )

                for definition in definitions:
                    outer_rep = int(definition["outer_rep"])
                    direction = str(definition["direction"])

                    expected = {
                        (
                            float(row.fraction_requested),
                            int(row.top_m),
                        )
                        for row in mapping.itertuples(index=False)
                    }
                    existing = set()

                    if len(fraction_run_df):
                        current = fraction_run_df[
                            (fraction_run_df["fit_fingerprint"].astype(str) == fingerprint)
                            & (fraction_run_df["condition"].astype(str) == condition)
                            & (fraction_run_df["K"].astype(int) == int(K))
                            & (fraction_run_df["outer_rep"].astype(int) == outer_rep)
                            & (fraction_run_df["direction"].astype(str) == direction)
                        ]
                        existing = {
                            (
                                float(row.fraction_requested),
                                int(row.top_m),
                            )
                            for row in current.itertuples(index=False)
                        }

                    missing = expected - existing
                    if (
                        not missing
                        and SKIP_EXISTING
                        and not FORCE_RECOMPUTE
                    ):
                        continue

                    ranked = ranked_for_direction(
                        fingerprint,
                        condition,
                        K,
                        outer_rep,
                        direction,
                    )
                    evaluation_subjects = [
                        subjects[int(index)]
                        for index in definition["evaluation_idx"]
                    ]
                    missing_m = sorted({
                        int(top_m)
                        for _, top_m in missing
                    })

                    evaluated = evaluate_reconstruction_topm_fast(
                        evaluation_subjects,
                        ranked,
                        missing_m,
                        NREPS_DECODE,
                        evaluation_seed_for(
                            K,
                            outer_rep,
                            definition["direction_offset"],
                        ),
                    )

                    rows = []
                    for map_row in mapping.itertuples(index=False):
                        pair = (
                            float(map_row.fraction_requested),
                            int(map_row.top_m),
                        )
                        if pair not in missing:
                            continue

                        source = evaluated[
                            evaluated["top_m"].astype(int)
                            == int(map_row.top_m)
                        ]
                        if len(source) != 1:
                            raise RuntimeError(
                                f"Missing K={K}, m={map_row.top_m}"
                            )

                        output = source.iloc[0].to_dict()
                        output.update({
                            "fit_label": fit_label,
                            "fit_fingerprint": fingerprint,
                            "fit_path": str(fit["path"]),
                            "condition": condition,
                            "K": int(K),
                            "outer_rep": outer_rep,
                            "direction": direction,
                            "outer_seed": int(definition["outer_seed"]),
                            "fraction_requested": float(
                                map_row.fraction_requested
                            ),
                            "actual_fraction": float(
                                map_row.actual_fraction
                            ),
                        })
                        rows.append(output)

                    fraction_run_df = append_dedupe(
                        fraction_run_df,
                        pd.DataFrame(rows),
                        FRACTION_RUN_KEYS,
                    )
                    save_csv(fraction_run_df, FRACTION_RUN_CSV)

                    print(
                        f"  {condition} | K={K} | outer "
                        f"{outer_rep + 1}/{N_OUTER_REPS} | "
                        f"{direction} | newly computed m={missing_m}",
                        flush=True,
                    )

                    del evaluation_subjects
                    gc.collect()

            del fit
            gc.collect()
else:
    print(
        "Matched-fraction analysis disabled. Enable it after validating "
        "the K=14 local output."
    )
````

## 2. Define the temporal archetype generator anchors

For temporal MS-AA, the shared generator matrix \(C\) has one row per TR and
one column per archetype. The primary anchor for archetype \(k\) is
\(\arg\max_t C_{tk}\). The table below also reports the maximum generator
weight and an effective number of generator TRs, so the anchor concentration is
explicit rather than assumed.


````python
# ============================================================
# GENERATOR-ANCHOR TABLE AND COMMON MASKS
# ============================================================

ANCHOR_TABLE_CSV = OUTPUT_DIR / "K14_generator_anchor_TRs.csv"
ANCHOR_MASK_TABLE_CSV = OUTPUT_DIR / "K14_generator_anchor_masks.csv"

k14_fit_row = k14_fit_table.iloc[0]
k14_fit = load_fit(k14_fit_row["fit_path"])
k14_fingerprint = str(k14_fit_row["fingerprint"])
k14_fit_label = str(k14_fit_row["fit_label"])

C = np.asarray(k14_fit["C"], dtype=float)

if C.ndim != 2:
    raise ValueError(f"Expected a 2D generator matrix C, found {C.shape}.")

if C.shape[1] != REFERENCE_K and C.shape[0] == REFERENCE_K:
    C = C.T

if C.shape[1] != REFERENCE_K:
    raise ValueError(
        f"Expected C to have {REFERENCE_K} columns; found {C.shape}."
    )

T_GENERATOR = int(C.shape[0])

anchor_rows = []

for archetype in range(REFERENCE_K):
    weights = np.clip(C[:, archetype], 0.0, None)
    total = float(weights.sum())
    probabilities = weights / total if total > 0 else np.zeros_like(weights)
    effective_n = (
        float(1.0 / np.sum(probabilities ** 2))
        if np.sum(probabilities ** 2) > 0
        else np.nan
    )
    order = np.argsort(weights)[::-1]

    anchor_rows.append({
        "archetype": int(archetype),
        "anchor_TR": int(order[0]),
        "max_generator_weight": float(weights[order[0]]),
        "second_TR": int(order[1]) if len(order) > 1 else np.nan,
        "second_generator_weight": (
            float(weights[order[1]]) if len(order) > 1 else np.nan
        ),
        "effective_generator_TRs": effective_n,
        "generator_column_sum": total,
    })

anchor_df = pd.DataFrame(anchor_rows)
save_csv(anchor_df, ANCHOR_TABLE_CSV)
display(anchor_df)


def anchor_keep_mask(n_timepoints, half_window):
    excluded = np.zeros(int(n_timepoints), dtype=bool)

    for anchor in anchor_df["anchor_TR"].astype(int):
        start = max(0, int(anchor) - int(half_window))
        stop = min(int(n_timepoints), int(anchor) + int(half_window) + 1)
        excluded[start:stop] = True

    return ~excluded


mask_rows = []
ANCHOR_KEEP_MASKS = {}

for half_window in sorted(set(map(int, ANCHOR_HALF_WINDOWS))):
    keep = anchor_keep_mask(T_GENERATOR, half_window)
    ANCHOR_KEEP_MASKS[int(half_window)] = keep

    excluded_trs = np.where(~keep)[0].astype(int).tolist()

    mask_rows.append({
        "anchor_half_window": int(half_window),
        "n_total_TRs": int(T_GENERATOR),
        "n_excluded_TRs": int((~keep).sum()),
        "n_remaining_TRs": int(keep.sum()),
        "masked_chance": float(1.0 / keep.sum()),
        "excluded_TRs": str(excluded_trs),
    })

anchor_mask_df = pd.DataFrame(mask_rows)
save_csv(anchor_mask_df, ANCHOR_MASK_TABLE_CSV)
display(anchor_mask_df)
````

## 3. Frozen-ranking, anchor-masked held-out reconstruction decoding

The participant-held-out ranking from the original unmasked analysis is held
fixed. On the evaluation half only, all generator-anchor windows are removed
before temporal normalization and timepoint matching. This avoids giving the
masked control a new selection advantage and asks whether the original
selected subsets generalize to non-anchor TRs.


````python
# ============================================================
# ANCHOR-MASKED DECODER
# ============================================================

ANCHOR_MASK_RUN_CSV = OUTPUT_DIR / "K14_anchor_masked_runs.csv"
ANCHOR_MASK_SUMMARY_CSV = OUTPUT_DIR / "K14_anchor_masked_summary.csv"
ANCHOR_MASK_COMPARISON_CSV = (
    OUTPUT_DIR / "K14_anchor_masked_vs_unmasked.csv"
)

anchor_mask_run_df = load_csv(ANCHOR_MASK_RUN_CSV)

ANCHOR_MASK_RUN_KEYS = [
    "fit_fingerprint",
    "condition",
    "outer_rep",
    "direction",
    "anchor_half_window",
    "top_m",
    "ranking_mode",
]


def raw_full_temporal_reconstruction_stack(subjects):
    return np.stack(
        [
            (
                np.asarray(subject["sXC"], dtype=float)
                @ np.asarray(subject["S"], dtype=float)
            ).T
            for subject in subjects
        ],
        axis=0,
    )


def masked_normalized_stack(raw_stack, keep_mask):
    kept = np.asarray(raw_stack, dtype=float)[:, keep_mask, :]
    return normalize_stack_over_time(kept)


def evaluate_anchor_masked_topm(
    subjects,
    ranked,
    top_m_values,
    keep_mask,
    half_window,
    nreps,
    seed,
):
    K = len(ranked)
    desired = sorted({
        int(m) for m in top_m_values
        if 1 <= int(m) <= K
    })

    if not desired:
        return pd.DataFrame()

    full_raw = raw_full_temporal_reconstruction_stack(subjects)
    full_stack = masked_normalized_stack(full_raw, keep_mask)
    full_summary = batch_decode_stack(full_stack, nreps=nreps, seed=seed)

    T_remaining = int(keep_mask.sum())
    chance = float(1.0 / T_remaining)
    n_subjects, T_full, V = full_raw.shape
    cumulative_raw = np.zeros((n_subjects, T_full, V), dtype=float)

    rows = []
    checkpoints = set(desired)

    for position, archetype in enumerate(ranked[:max(desired)], start=1):
        cumulative_raw += temporal_contribution_stack(subjects, archetype)

        if position not in checkpoints:
            continue

        if position == K:
            summary = full_summary
            retention = 1.0
        else:
            masked_stack = masked_normalized_stack(cumulative_raw, keep_mask)
            summary = batch_decode_stack(
                masked_stack,
                nreps=nreps,
                seed=seed,
            )
            denominator = full_summary["mean"] - chance
            retention = (
                (summary["mean"] - chance) / denominator
                if abs(denominator) > 1e-12
                else np.nan
            )

        rows.append({
            "space": "reconstruction_anchor_masked",
            "ranking_mode": ANCHOR_RANKING_MODE,
            "anchor_half_window": int(half_window),
            "n_total_TRs": int(T_full),
            "n_excluded_TRs": int((~keep_mask).sum()),
            "n_remaining_TRs": int(T_remaining),
            "top_m": int(position),
            "selected_archetypes": str([
                int(value) for value in ranked[:position]
            ]),
            "mean": summary["mean"],
            "err": summary["err"],
            "std": summary["std"],
            "count": summary["count"],
            "rank_mean": summary["rank_mean"],
            "error_mean": summary["error_mean"],
            "full_mean_same_eval": full_summary["mean"],
            "full_err_same_eval": full_summary["err"],
            "retention_above_chance": retention,
            "chance": chance,
        })

    return pd.DataFrame(rows)
````

````python
# ============================================================
# RUN / RESUME ANCHOR-MASKED K=14 ANALYSIS
# ============================================================

if RUN_ANCHOR_MASKED:
    for condition in CONDITIONS:
        definitions, subjects = ensure_rankings(
            k14_fit,
            k14_fit_label,
            k14_fingerprint,
            condition,
        )

        for definition in definitions:
            outer_rep = int(definition["outer_rep"])
            direction = str(definition["direction"])

            ranked = ranked_for_direction(
                k14_fingerprint,
                condition,
                REFERENCE_K,
                outer_rep,
                direction,
            )

            evaluation_subjects = [
                subjects[int(index)]
                for index in definition["evaluation_idx"]
            ]

            evaluation_seed = evaluation_seed_for(
                REFERENCE_K,
                outer_rep,
                definition["direction_offset"],
            )

            for half_window, keep_mask in ANCHOR_KEEP_MASKS.items():
                existing_m = set()

                if len(anchor_mask_run_df):
                    current = anchor_mask_run_df[
                        (anchor_mask_run_df["fit_fingerprint"].astype(str) == k14_fingerprint)
                        & (anchor_mask_run_df["condition"].astype(str) == condition)
                        & (anchor_mask_run_df["outer_rep"].astype(int) == outer_rep)
                        & (anchor_mask_run_df["direction"].astype(str) == direction)
                        & (anchor_mask_run_df["anchor_half_window"].astype(int) == int(half_window))
                        & (anchor_mask_run_df["ranking_mode"].astype(str) == ANCHOR_RANKING_MODE)
                    ]
                    existing_m = set(current["top_m"].astype(int))

                missing_m = sorted(
                    set(map(int, ANCHOR_MASK_TOP_M_VALUES)) - existing_m
                )

                if (
                    not missing_m
                    and SKIP_EXISTING
                    and not FORCE_RECOMPUTE
                ):
                    continue

                evaluated = evaluate_anchor_masked_topm(
                    evaluation_subjects,
                    ranked,
                    missing_m,
                    keep_mask,
                    half_window,
                    NREPS_DECODE,
                    evaluation_seed,
                )

                if len(evaluated):
                    evaluated["fit_label"] = k14_fit_label
                    evaluated["fit_fingerprint"] = k14_fingerprint
                    evaluated["fit_path"] = str(k14_fit["path"])
                    evaluated["condition"] = condition
                    evaluated["K"] = REFERENCE_K
                    evaluated["outer_rep"] = outer_rep
                    evaluated["direction"] = direction
                    evaluated["outer_seed"] = int(definition["outer_seed"])
                    evaluated["n_ranking_subjects"] = int(
                        len(definition["ranking_idx"])
                    )
                    evaluated["n_evaluation_subjects"] = int(
                        len(definition["evaluation_idx"])
                    )

                    anchor_mask_run_df = append_dedupe(
                        anchor_mask_run_df,
                        evaluated,
                        ANCHOR_MASK_RUN_KEYS,
                    )
                    save_csv(anchor_mask_run_df, ANCHOR_MASK_RUN_CSV)

                print(
                    f"  masked | {condition} | outer {outer_rep + 1}/"
                    f"{N_OUTER_REPS} | {direction} | +/-{half_window} TR | "
                    f"new m={missing_m}",
                    flush=True,
                )

            del evaluation_subjects
            gc.collect()
else:
    print("Anchor-masked analysis disabled.")
````

## 4. Compare masked and unmasked accumulation curves

Directions are averaged within each outer split before uncertainty is computed
across the 20 outer splits. The comparison table includes absolute accuracy,
above-chance retention, and the masked-minus-unmasked change at each \(m\).


````python
# ============================================================
# SUMMARIZE MASKED RESULTS AND COMPARE WITH REVISION 10A
# ============================================================

anchor_mask_run_df = load_csv(ANCHOR_MASK_RUN_CSV)
unmasked_run_df = load_csv(UNMASKED_K14_RUN_CSV)


def summarize_outer_runs(
    runs,
    grouping_extra=None,
    accuracy_column="mean",
    retention_column="retention_above_chance",
):
    grouping_extra = list(grouping_extra or [])

    outer = (
        runs
        .groupby(
            [
                "condition",
                "K",
                "top_m",
                *grouping_extra,
                "outer_rep",
            ],
            as_index=False,
        )
        .agg(
            accuracy=(accuracy_column, "mean"),
            retention=(retention_column, "mean"),
            chance=("chance", "mean"),
            full_accuracy=("full_mean_same_eval", "mean"),
        )
    )

    summary = (
        outer
        .groupby(
            [
                "condition",
                "K",
                "top_m",
                *grouping_extra,
            ],
            as_index=False,
        )
        .agg(
            mean_accuracy=("accuracy", "mean"),
            outer_sem_accuracy=("accuracy", sem),
            mean_retention=("retention", "mean"),
            outer_sem_retention=("retention", sem),
            mean_chance=("chance", "mean"),
            mean_full_accuracy=("full_accuracy", "mean"),
            n_outer=("outer_rep", "nunique"),
        )
    )

    return outer, summary


if len(anchor_mask_run_df):
    _, anchor_mask_summary_df = summarize_outer_runs(
        anchor_mask_run_df,
        grouping_extra=["anchor_half_window", "ranking_mode"],
    )
    save_csv(anchor_mask_summary_df, ANCHOR_MASK_SUMMARY_CSV)
    display(anchor_mask_summary_df.head(30))
else:
    anchor_mask_summary_df = pd.DataFrame()
    print("No anchor-masked rows are available yet.")


if len(unmasked_run_df):
    unmasked_reconstruction = unmasked_run_df[
        unmasked_run_df["space"].astype(str) == "reconstruction"
    ].copy()
    _, unmasked_summary_df = summarize_outer_runs(
        unmasked_reconstruction,
    )
else:
    unmasked_summary_df = pd.DataFrame()
    print(
        "Unmasked K=14 runs were not found at "
        f"{UNMASKED_K14_RUN_CSV}. Run Revision 10A first."
    )


comparison_rows = []

if len(anchor_mask_summary_df) and len(unmasked_summary_df):
    for half_window in sorted(
        anchor_mask_summary_df["anchor_half_window"].astype(int).unique()
    ):
        masked = anchor_mask_summary_df[
            anchor_mask_summary_df["anchor_half_window"].astype(int)
            == int(half_window)
        ].copy()

        merged = masked.merge(
            unmasked_summary_df,
            on=["condition", "K", "top_m"],
            suffixes=("_masked", "_unmasked"),
            how="left",
        )

        merged["accuracy_delta_masked_minus_unmasked"] = (
            merged["mean_accuracy_masked"]
            - merged["mean_accuracy_unmasked"]
        )
        merged["retention_delta_masked_minus_unmasked"] = (
            merged["mean_retention_masked"]
            - merged["mean_retention_unmasked"]
        )
        merged["above_chance_ratio_masked_to_unmasked"] = (
            (
                merged["mean_accuracy_masked"]
                - merged["mean_chance_masked"]
            )
            /
            (
                merged["mean_accuracy_unmasked"]
                - merged["mean_chance_unmasked"]
            ).replace(0, np.nan)
        )

        comparison_rows.append(merged)

    anchor_comparison_df = pd.concat(
        comparison_rows,
        ignore_index=True,
    )
    save_csv(anchor_comparison_df, ANCHOR_MASK_COMPARISON_CSV)
    display(anchor_comparison_df.head(30))
else:
    anchor_comparison_df = pd.DataFrame()
````

````python
# ============================================================
# MASKED VS UNMASKED FIGURES
# ============================================================

if len(anchor_comparison_df):
    for condition in CONDITIONS:
        for half_window in sorted(
            anchor_comparison_df["anchor_half_window"].astype(int).unique()
        ):
            current = (
                anchor_comparison_df[
                    (anchor_comparison_df["condition"] == condition)
                    & (
                        anchor_comparison_df["anchor_half_window"].astype(int)
                        == int(half_window)
                    )
                ]
                .sort_values("top_m")
            )

            if len(current) == 0:
                continue

            plt.figure(figsize=(7.5, 4.8))
            plt.plot(
                current["top_m"],
                current["mean_accuracy_unmasked"],
                marker="o",
                linewidth=2,
                label="Unmasked",
            )
            plt.fill_between(
                current["top_m"],
                current["mean_accuracy_unmasked"]
                - current["outer_sem_accuracy_unmasked"].fillna(0),
                current["mean_accuracy_unmasked"]
                + current["outer_sem_accuracy_unmasked"].fillna(0),
                alpha=0.16,
            )
            plt.plot(
                current["top_m"],
                current["mean_accuracy_masked"],
                marker="o",
                linewidth=2,
                label=f"Anchor masked (+/-{half_window} TR)",
            )
            plt.fill_between(
                current["top_m"],
                current["mean_accuracy_masked"]
                - current["outer_sem_accuracy_masked"].fillna(0),
                current["mean_accuracy_masked"]
                + current["outer_sem_accuracy_masked"].fillna(0),
                alpha=0.16,
            )
            plt.axhline(
                current["mean_chance_masked"].iloc[0],
                linestyle="--",
                linewidth=1,
                label="Masked chance",
            )
            plt.xlabel("Number of selected archetypes (m)")
            plt.ylabel("Held-out timepoint decoding accuracy")
            plt.title(
                f"K=14 temporal MS-AA | {condition} | "
                f"generator-anchor masking"
            )
            plt.legend(frameon=False)
            plt.tight_layout()
            plt.savefig(
                OUTPUT_DIR
                / f"K14_{condition}_anchor_mask_pm{half_window}_accuracy.png",
                dpi=300,
                bbox_inches="tight",
            )
            plt.show()

            plt.figure(figsize=(7.5, 4.8))
            plt.plot(
                current["top_m"],
                current["mean_retention_unmasked"],
                marker="o",
                linewidth=2,
                label="Unmasked",
            )
            plt.plot(
                current["top_m"],
                current["mean_retention_masked"],
                marker="o",
                linewidth=2,
                label=f"Anchor masked (+/-{half_window} TR)",
            )
            for threshold in [0.50, 0.80, 0.90, 1.00]:
                plt.axhline(threshold, linestyle="--", linewidth=0.8)
            plt.xlabel("Number of selected archetypes (m)")
            plt.ylabel(
                "Fraction of above-chance full-model performance retained"
            )
            plt.title(
                f"K=14 temporal MS-AA | {condition} | "
                f"masked retention"
            )
            plt.legend(frameon=False)
            plt.tight_layout()
            plt.savefig(
                OUTPUT_DIR
                / f"K14_{condition}_anchor_mask_pm{half_window}_retention.png",
                dpi=300,
                bbox_inches="tight",
            )
            plt.show()
````

Interpretation:

- If the low-\(K\) advantage persists when the retained fraction \(m/K\) is
  matched, this supports a particularly favorable compact temporal scale.
- If it attenuates, the fixed top-10 peak is substantially explained by
  retaining a larger fraction of the small \(K=14\) model.
- If anchor-masked accuracy remains above the new chance level and the
  accumulation curve persists, selected spatial archetypes carry information
  beyond the TRs that explicitly generate them.
- If masked performance collapses, the temporal top-\(m\) result is primarily
  an exemplar- or anchor-timepoint effect.
- A partial reduction supports a mixed account: anchor TRs contribute strongly,
  but distributed non-anchor information remains.
